```python
# Nama        : Kasih Nuraini
# NIM         : 4232501003
# Kelas       : 3A Pagi
# Mata Kuliah : Pemrograman Dasar

# SOAL NOMOR 1, Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko
# memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

jumlah_buku = 7
harga = 18500
diskon = 0.1
uang = 150000
orang = 4

#1 Total belanja, besar diskon, dan jumlah yang harus dibayar
total = jumlah_buku * harga
diskon = total * diskon
bayar = total - diskon

#2 Uang kembalian.
kembalian = uang - bayar

# 3 Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat,
# berapa iuran per orang dan berapa rupiah sisanya?
iuran = bayar // orang
sisa = bayar % orang

#4 Apakah uang Budi cukup?
uang_cukup = uang >= total

print("Total (Rp)          : ", total)
print("Diskon (Rp)         : ", diskon)
print("Bayar (Rp)          : ", bayar)
print("Kembalian (Rp)      : ", kembalian)
print("Iuran per orang (Rp): ", iuran)
print("Sisa                : ", sisa)
print("Uang cukup?         : ", uang_cukup)
```

```python
# Nama        : Kasih Nuraini
# NIM         : 4232501003
# Kelas       : 3A Pagi
# Mata Kuliah : Pemrograman Dasar

# SOAL NOMOR 2, Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel.
# Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8).
# Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.

jumlah_panel = 12
daya = 0.45   #450 Wp = 0.45 kW
ratarata_penyinaran = 5
efisiensi = 0.8
kebutuhan_listrik_perhari = 25
tarif_listrik = 1444.70

#1 Energi yang dihasilkan per hari (kWh)
energi_per_hari = jumlah_panel * daya * ratarata_penyinaran * efisiensi

#2 Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan
sebulan = 30
energi_per_bulan = energi_per_hari * sebulan
nilai_rupiah = tarif_listrik * energi_per_bulan

#3 Kekurangan energi perhari terhadap kebutuhan
kekurangan = kebutuhan_listrik_perhari - energi_per_hari

#4 apakah PLTS sudah mencukup kebutuhan rumah?
mencukupi = energi_per_hari >= kebutuhan_listrik_perhari

print("Energi per hari (kWh)  : ", energi_per_hari)
print("Energi per bulan (kWh) : ", energi_per_bulan)
print("Nilai listrik (Rp)     : ", nilai_rupiah)
print("Kekurangan (kWh)       : ", kekurangan)
print("Mencukupi?             : ", mencukupi)
```

```python
# Nama        : Kasih Nuraini
# NIM         : 4232501003
# Kelas       : 3A Pagi
# Mata Kuliah : Pemrograman Dasar

# SOAL NOMOR 3, Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi
# bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas
# tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

daya_generator = 200
waktu_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus_kabel = 20
hambatan_kabel = 0.5

#1 Energi yang dibangkitkan per hari (kWh)
energi = daya_generator * waktu_operasi

#2 Kebutuhan solar per hari dan biayanya
kebutuhan_solar = energi * konsumsi_solar
biaya = harga_solar * kebutuhan_solar

#3 Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
hari = kapasitas_tangki // kebutuhan_solar
sisa_solar = kapasitas_tangki - (kebutuhan_solar * hari)

#4 Rugi daya pada kabel (P = I² × R)
rugi_daya = arus_kabel * arus_kabel * hambatan_kabel

#5 Apakah rugi kabel tergolong tinggi, yaitu lebih dari 150 W dan arus minimal 20 A?
rugi_tinggi = rugi_daya > 150 and arus_kabel >= 20

print("Energi /hari (kWh)             : ", energi)
print("Kebutuhan solar /hari (liter)  : ", kebutuhan_solar)
print("Biaya solar (Rp)               : ", biaya)
print("Tangki penuh cukup untuk (hari): ", hari)
print("Sisa solar (liter)             : ", sisa_solar)
print("Rugi kabel (W)                 : ", rugi_daya)
print("Rugi tinggi?                   : ", rugi_tinggi)
```
