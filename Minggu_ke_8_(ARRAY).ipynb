```python
# AKTIVITAS 6D
suhu = [
    61, 60, 60, 59, 59, 60,   # pukul 00.00–05.00
    63, 66, 69, 71, 72, 73,   # pukul 06.00–11.00
    74, 74, 75, 74, 73, 72,   # pukul 12.00–17.00
    70, 68, 66, 64, 63, 62    # pukul 18.00–23.00
]
# Mengurutkan data suhu
urut = sorted(suhu)
print("3 terendah :", urut[:3])
print("3 tertinggi:", urut[-3:])
print("Data asli  :", suhu[:3])

# Menghitung median
median = (urut[11] + urut[12]) / 2
print("Median     :", median, "°C")

# Menyalin dan mengurutkan data dari terbesar
salinan = suhu.copy()
salinan.sort(reverse=True)
print("Terbesar dulu:", salinan[:5])

# Tebak dulu, baru jalankan
print("suhu[6:12:2] :", suhu[6:12:2])                     # [63, 69, 72]
print("len(suhu[::3]) :", len(suhu[::3]))                 # 8
print("suhu[-1] - suhu[0] :", suhu[-1] - suhu[0])         # 1
print("sorted(suhu)[0] :", sorted(suhu)[0])               # 59
```

```python
# AKTIVITAS 6F
skor = [80, 65, 90, 75, 70]
total = 0
jumlah_lulus = 0
for nilai in skor:
    total = total + nilai
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)
rata_rata = total / len(skor)

print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
# AKTIVITAS 6G
suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 5 hari
rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

```python
# AKTIVITAS 6H
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

hari = list(range(1, 31))
jam_operasi = [24] * 30
jam_operasi[16] = 18
NAMA_HARI = ["Min", "Sen", "Sel", "Rab", "Kam", "Jum", "Sab"]

# Memeriksa perubahan energi harian
for i in range(1, len(energi)):
    perubahan = (energi[i] - energi[i - 1]) / energi[i - 1] * 100
    if perubahan < -20:
        print(f"Hari {i + 1}: {perubahan:+.1f}% dibanding hari sebelumnya, perlu diselidiki")

# Memeriksa jam operasi
for i in range(len(energi)):
    if jam_operasi[i] < 24:
        print("Hari", hari[i], ":", jam_operasi[i], "jam,", energi[i], "MWh")

# Memisahkan hari kerja dan akhir pekan
kerja = []
libur = []
for i in range(len(energi)):
    if NAMA_HARI[i % 7] == "Sab" or NAMA_HARI[i % 7] == "Min":
        libur.append(energi[i])
    else:
        kerja.append(energi[i])

print("Rata-rata hari kerja :", round(sum(kerja) / len(kerja)), "MWh")
print("Rata-rata akhir pekan:", round(sum(libur) / len(libur)), "MWh")
```

```python
# AKTIVITAS 6I
# Membuat list suhu selama 7 hari
suhu = [32, 35, 38, 34, 37, 31, 40]

print("Data suhu awal:", suhu)

# 1. Menampilkan 3 hari pertama dan 2 hari terakhir
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# 2. Menambahkan dan mengeluarkan satu data
suhu.append(36)
print("Setelah append:", suhu)

suhu.pop()
print("Setelah pop:", suhu)

# 3. Mengurutkan data tanpa mengubah list asli
urut = sorted(suhu)
print("Data suhu terurut:", urut)
print("Data suhu asli:", suhu)

# 4. Menghitung rata-rata dan mencari suhu tertinggi
rata_rata = sum(suhu) / len(suhu)
tertinggi = max(suhu)
hari_tertinggi = suhu.index(tertinggi) + 1

print("Rata-rata suhu:", rata_rata)
print("Suhu tertinggi:", tertinggi, "°C")
print("Suhu tertinggi terdapat pada hari ke-", hari_tertinggi)

# 5. Membuat list suhu di atas rata-rata
di_atas_rata = []

for nilai in suhu:
    if nilai > rata_rata:
        di_atas_rata.append(nilai)

print("Suhu di atas rata-rata:", di_atas_rata)
```

```python
# AKTIVITAS 6J
nilai = [
    [80, 70, 90],     # Ana
    [65, 75, 60],     # Budi
    [90, 85, 95],     # Cici
]

print("Baris Budi:", nilai[1])
print("Ujian 3 Budi:", nilai[1][2])
print("Skor ujian 2 Cici:", nilai[2][1])
print("Jumlah baris:", len(nilai))
print("Jumlah kolom:", len(nilai[0]))

# Total nilai setiap mahasiswa
for b in range(len(nilai)):
    print("Mahasiswa", b + 1, "total:", sum(nilai[b]))

# Mengambil skor ujian pertama semua mahasiswa
ujian1 = []
for baris in nilai:
    ujian1.append(baris[0])

print("Skor ujian 1:", ujian1)
```
