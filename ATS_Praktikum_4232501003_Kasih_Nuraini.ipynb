```python
# SOAL 1

tegangan = 220
arus = 15

# a. daya listrik
daya_listrik = tegangan * arus

print(f"Tegangan: {tegangan} Volt")
print(f"Arus: {arus} Ampere")
print(f"Daya listrik: {daya_listrik} Watt")

# b. efisiensi eurbin
energi_input = 500
energi_output = 400
efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi turbin: {efisiensi:.2f}%")
```

```python
# SOAL 2

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# a. Menghitung Energi Harian PLTD
energi = daya_kw * jam

print(f"Daya PLTD: {daya_kw} kW")
print(f"Lama operasi: {jam} jam")
print(f"Energi harian: {energi} kWh")

# b. capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

print(f"Capacity factor: {capacity_factor:.2f}%")
```

```python
# SOAL 9
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0
for i, nilai in enumerate(data, start=1):
    try:
        vibrasi = float(nilai)
    except ValueError:
        print(f"Unit {i}: data tidak valid")
        jumlah_tidak_valid += 1
        continue
    if vibrasi > 2.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")
    if vibrasi > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

```python
# SOAL 10
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0
for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1
    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break
    if suhu > 70.0:
        print(f"Warning: Titik {i} suhu bearing tinggi!")
print(f"Jumlah data valid: {jumlah_valid}")
```

```python
# Program Pemantauan Konsumsi Listrik Harian

data_listrik = [110, 120, 160, 240, 310, 500]
total_energi = 0
jumlah_data = 0

# Konsumsi listrik setiap hari
for i, energi in enumerate(data_listrik, start=1):
    try:
        energi = float(energi)
        total_energi += energi
        jumlah_data += 1
        print(f"Hari ke-{i}: {energi} kWh")

        # kondisi konsumsi listrik
        if energi > 200:
            print("Status: Konsumsi sangat tinggi!")
        elif energi > 150:
            print("Status: Konsumsi tinggi.")
        else:
            print("Status: Konsumsi normal.")

    except ValueError:
        print(f"Hari ke-{i}: Data tidak valid!")

# Rata-rata konsumsi listrik
if jumlah_data > 0:
    rata_rata = total_energi / jumlah_data
    print()
    print("HASIL PEMANTAUAN")
    print(f"Total energi: {total_energi:.2f} kWh")
    print(f"Rata-rata konsumsi: {rata_rata:.2f} kWh")
    print(f"Jumlah data valid: {jumlah_data}")
else:
    print("Tidak ada data konsumsi listrik yang valid.")
```
