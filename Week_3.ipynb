1. Kenapa Ekspresi 10 + 5 * 2 Menghasilkan 20, Bukan 30?

---
   Aturan kerja di Python mengikuti urutan matematika dasar. Ini saya dapatkan dari artikel Duniailkom, yang menjelaskan mengenai urutan prioritas operator (operator precedence).

   Karena operator perkalian (*) punya tingkatan prioritas yang lebih tinggi daripada penjumlahan (+), Python tidak langsung menghitung dari kiri ke kanan. Saat membaca baris kode 10 + 5 * 2, Python bakal mengeksekusi 5 * 2 terlebih dahulu sampai ketemu hasil 10. Baru setelah itu hasilnya dijumlahkan dengan angka 10 yang di depan. Makanya hasil akhirnya ketemu 20, bukan 30.

   Kalau mau penambahannya dihitung duluan, kita tinggal bungkus penambahannya pakai tanda kurung jadi (10 + 5) * 2, karena tanda kurung selalu punya prioritas paling tinggi

2. Perbedaan Operator Relasional dan Operator Logika

---

   Kalau dari yang saya pelajari di artikel Duniailkom dan Petani Kode, bedanya sederhana:

   Operator Relasional (>, <, ==, !=, dll.) itu tugasnya cuma buat membandingkan dua nilai. Hasilnya pasti cuma dua kemungkinan: benar (True) atau salah (False).

   Operator Logika (and, or, not) tugasnya buat menggabungkan beberapa kondisi tadi jadi satu keputusan akhir.

   Contoh di kehidupan nyata: Misalkan sistem pembayaran otomatis di e-wallet atau m-banking. Transaksi cuma bakal diproses kalau nominal transfer tidak melebihi saldo dan status akun pengguna masih aktif.

Kalau ditulis dalam bentuk kode:


```python
# Pengecekan: apakah 50000 <= 100000 DAN "Aktif" == "Aktif"
izinkan_transaksi = (50000 <= 100000) and ("Aktif" == "Aktif")

print(izinkan_transaksi)  # Hasilnya: True
```

3. Kasus bug atau kesalahan perhitungan yang terjadi akibat salah memahami operator precedence

---
   Salah satu kasus bug (kesalahan logika) yang sering ditemui di forum diskusi programmer terjadi saat seseorang mau menghitung nilai rata-rata dari dua variabel, misalnya angka 10 dan 20. Niatnya ingin menghitung (10 + 20) / 2 = 15, tapi di dalam kode dituliskan langsung tanpa tanda kurung atau rata_rata = 10 + 20 / 2.
   
   Karena operator pembagian (/) punya tingkatan prioritas yang lebih tinggi daripada penjumlahan (+), Python bakal mengerjakan 20 / 2 terlebih dahulu yang hasilnya 10. Setelah itu baru ditambahkan dengan angka 10 di depan (10 + 10). Akibatnya, program malah mengeluarkan hasil 20, bukannya 15.
   
   Perbaikan Kode:Masalah ini diselesaikan dengan membungkus operasi penjumlahan menggunakan tanda kurung (). Tanda kurung memaksa Python buat ngerjain penjumlahannya dulu sebelum dibagi:Python

```python
# Kode Salah (Yang menghasilkan Bug):
rata_rata_salah = 10 + 20 / 2
print(f"Hasil kode yang salah (tanpa kurung) : {rata_rata_salah}")  # Output: 20.0

# Kode Benar (Hasil Rata-Rata Seharusnya):
rata_rata_benar = (10 + 20) / 2
print(f"Hasil kode yang benar (pakai kurung) : {rata_rata_benar}")  # Output: 15.0
```
